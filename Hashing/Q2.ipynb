Frequency map / Dictionary

```python
## Using freq_map

nums = [5, 6, 7, 7, 1, 9, 1, 1, 5, 1, 1]

freq_map = {}

for i in range(len(nums)):

    if nums[i] in freq_map:
        freq_map[nums[i]] += 1
    else:
        freq_map[nums[i]] = 1

print(freq_map)
print(freq_map[1])
```

```python
## Using hash_map

nums = [5, 6, 7, 7, 1, 9, 1, 1, 5, 1, 1]

hash_map = {}
n = len(nums)

for i in range(0, n):
    hash_map[nums[i]] = hash_map.get(nums[i], 0) + 1

print(hash_map)
print(hash_map[5])     ## TC = O(N) , SC = O(N) Average TC = O(1)
```

```python
## hash_map using optimal solution         constraints:- 1<=n[i]<=10 
n = [ 5,3,2,2,1,5,5,7,5,10]
m = [ 1,111,10,8,9,6,5,67,1,5]

hash_list=[0]*11
for num in n:
    hash_list[num] += 1
for num in m:
    if num < 1 or num > 10:
        print(0)
    else:
        print(hash_list[num])       ## TC = O(n+m) , SC = O(1) 
```

```python
# Character Hashing
# Constraint: 'a' <= s[i] <= 'z'

s = "azyxyyzaaaa"
q = ["d", "a", "y", "x"]

hash_list = [0] * 26

for ch in s:
    ascii_val = ord(ch)
    index = ascii_val - 97
    hash_list[index] += 1

for ch in q:
    ascii_val = ord(ch)
    index = ascii_val - 97
    print(hash_list[index])        ## TC =O(n+m) , SC = O(1)
```
