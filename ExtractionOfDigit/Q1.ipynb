```python
## 1.print the digits

n = 5873
num = n
while num > 0:
    last_digit = num % 10
    print(last_digit)
    num = num//10
    
```

```python
## Reverse the digits

n = 1234
num = n
result = 0

while num > 0:
    ld = num % 10
    result = result * 10 + ld
    num = num // 10

print(result)
```

```python
## 2. count digits

n = 5438
num = n
count = 0
while num > 0:
    count += 1
    num = num//10
print(count)
# TC = O(log10(n))
# SC = o(1)
```

```python
## 3. count digit using log10
from math import *

def countDigits(num):
    return int(log10(num) + 1)

n = 5438
print(countDigits(n))
```

```python
## 4. check palindrome

def isPalindrome(n):
    num = n
    result = 0

    while num > 0:
        ld = num % 10
        result = (result * 10) + ld
        num = num // 10

    return n == result

n = 3223
print(isPalindrome(n))

# TC = O(Log10(N))
# SC = O(1)
```

```python
## 5.Armstrong number

def isArmstrong(n):
    num = n
    total = 0
    num_of_digit = len(str(n))

    while num > 0:
        ld = num % 10
        total = total + (ld ** num_of_digit)
        num = num // 10

    return total == n

n = 153
print(isArmstrong(n))

# TC = O(log10(N))
# SC = O(1)
```

```python
# 6. Print the Factors or Divisors of a Given Number
# Brute Force soln

def factors(num):
    result = []

    for i in range(1, num + 1):
        if num % i == 0:
            result.append(i)

    return result

print(factors(12))

# TC = O(N)
# SC = O(k) k would be the total number of factors
```

```python
# 7. Print the Factors or Divisors of a Given Number
# Better Soln

def factors(num):
    result = []

    for i in range(1, num//2 + 1):
        if num % i == 0:
            result.append(i)

    result.append(num)
    return result

print(factors(12))

# TC = O(N/2)=0(N)
# SC = O(k) 
```

```python
#8. Print the Factors or Divisors of a Given Number
# Optimal Solution

from math import sqrt

def factors(num):
    result = []

    for i in range(1, int(sqrt(num)) + 1):
        if num % i == 0:
            result.append(i)

            if num // i != i:
                result.append(num // i)

    result.sort()
    return result

num = 12
print(factors(num))

# TC = O(sqrt(N) + K log K)
# SC = O(K)
```
