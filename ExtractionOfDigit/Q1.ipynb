```python
## 2. count digits

n = 5438
num = n
count = 0
while num > 0:
    count += 1
    num = num//10
print(count)
# TC = O(log10(n))
# SC = o(1)
```

```python
## 3. count digit using log10
from math import *

def countDigits(num):
    return int(log10(num) + 1)

n = 5438
print(countDigits(n))
```

```python
## 4. check palindrome

def isPalindrome(n):
    num = n
    result = 0

    while num > 0:
        ld = num % 10
        result = (result * 10) + ld
        num = num // 10

    return n == result

n = 3223
print(isPalindrome(n))

# TC = O(Log10(N))
# 


```
